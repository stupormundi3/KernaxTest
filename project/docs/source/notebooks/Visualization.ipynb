# Gaussian processes examples


## Simple comparison of the behavior of Square Exponential, Matern, and Periodic Kernel

```python
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '..')
from kernax.stationary import SEKernel, Matern32Kernel, PeriodicKernel
import numpy as np
import GP

kernels = {
    'SquareExpo': SEKernel(1.0),
    'Matern': Matern32Kernel(1.0),
    'Periodic': PeriodicKernel(1.0, 1.0, 5.0)
}

X = np.linspace(0, 10, 100).reshape(-1, 1)
y = np.squeeze(X * np.sin(X))
rng = np.random.RandomState(1)
#We choose randomly the observations 
training_indices = rng.choice(np.arange(y.size), size=10, replace=False)
X_train, y_train = X[training_indices], y[training_indices]

X_plot = np.linspace(0, 10, 400).reshape(-1,1)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, kern) in zip(axes, kernels.items()):
    gp = GP.GP(kern, Alpha=1e-5, NormalizeObs=True).fit(X_train, y_train, num_iters=200)

    #Prediction of the points produced by X_Plot
    mu, std = gp.predict(X_plot, Return_std=True)
    ax.plot(X, y, 'k--', lw=1, label='Real function')
    ax.scatter(X_train, y_train, c='red', s=30)
    ax.plot(X_plot, mu, 'b-', label='Mean')
    #Normalization 
    ax.fill_between(X_plot.ravel(), mu-1.96*std, mu+1.96*std, alpha=0.3, color='yellow')
    ax.set_title(name)
    ax.legend(fontsize=8)
plt.tight_layout()

plt.show()
```

## Measuring the influence of the length scale parameter in the context of a Square Exponentiel kernel

```python
length_scales = [0.1,1.0,10.0]
plt.figure(figsize=(12,4))
y = np.squeeze(X * np.sin(X))
for i, ls in enumerate(length_scales, 1):
    plt.subplot(1,3,i)
    
    gp = GP.GP(SEKernel(ls)).fit(X_train, y_train, num_iters=200)
    mu, std = gp.predict(X_plot, Return_std=True)
    plt.plot(X, y, 'k--', lw=1)
    plt.scatter(X_train, y_train, c='red', s=30)
    plt.plot(X_plot, mu, 'b-')
    plt.fill_between(X_plot.ravel(), mu-1.96*std, mu+1.96*std, alpha=0.3)
    plt.title(f'ℓ = {ls}')
plt.tight_layout()
plt.show()
```

## Example of a gaussian process with its mean function and 5 samples, in the context of a Square Exponential kernel

```python
gp = GP.GP(SEKernel(1.0), Alpha=1e-5).fit(X_train, y_train)
mu, cov = gp.predict(X_plot, Return_cov=True)
L = np.linalg.cholesky(cov + 1e-4*np.eye(len(X_plot)))
samples = mu[:,None] + L @ np.random.randn(len(X_plot), 5)

plt.figure(figsize=(10,5))
plt.plot(X, y, 'k--', label='Real function')
plt.scatter(X_train, y_train, c='red', s=40, label='Observations')
plt.plot(X_plot, samples, lw=0.8, alpha=0.7)
plt.fill_between(X_plot.ravel(), mu-1.96*std, mu+1.96*std, alpha=0.2, color='yellow')
plt.legend()
plt.title("5 posterior samples")
plt.show()
```
